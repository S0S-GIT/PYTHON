```python
a = int(input("Ingresa el primer número: "))
b = int(input("Ingresa el segundo número: "))

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)
print("a == b:", a == b)
print("a != b:", a != b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
print("a < b:", a < b)
print("a > b:", a > b)
```

Act.5 Programa que solicita información del estudiante:

1. Pida el nombre del estudiante
2. Pida el año de nacimiento
3. Pida el mes de nacimiento
4. Pida el año actual
5. Pida el mes actual
6. Te imprima tu edad + el número de mes y el nombre

```python
from datetime import datetime
nombre =(input("Cual es tu nombre/s: "))
apellido_P =(input("Cual es tu apellido parterno: "))
apellido_M =(input("Cual es tu apellido materno: "))
def nom_com(nombre, apellido_P, apellido_M):
    print("Hola " + nombre.title() + " " + apellido_P.title() + " " + apellido_M.title())
Mes_NA =int(input("Cual es tu mes de nacimiento en numero= "))
Año_NA =int(input("Cual es tu año de nacimiento= "))

Fe_AC= datetime.now()
Año_AC= Fe_AC.year
Mes_AC= Fe_AC.month

if Mes_NA <= Mes_AC:
    edad = Año_AC - Año_NA
    mes = Mes_AC - Mes_NA
else:
    edad = Año_AC - Año_NA - 1
    mes = (12 - Mes_NA) + Mes_AC

m_a = (edad * 12) + mes

def años_e(edad, mes, m_a):
    print("Tú tienes " + str(edad) + " años con " + str(mes) + " meses y " + str(m_a) + " meses vividos.")

nom_com(nombre, apellido_P, apellido_M)
años_e(edad, mes, m_a)
```
