```python
radio= 20.6538
area= 3.1416 * radio **2
```

```python
print("para un radio de " + str(radio) + " cm, tenemos un area de " + str(area) + " cm^2")
```

```python
print("para radio de {} cm, tenemos un área de {} cm^2" .format(radio,area))
```

```python
print("para radio de {: .2f} cm, tenemos un área de {: .2f} cm^2" .format(radio,area))
```

```python
print(f"para radio de {radio} cm, tenemos un área de {area} cm^2" .format(radio,area))
```

```python
print(f"para radio de {radio: .2f} cm, tenemos un área de {area: .2f} cm^2" .format(radio,area))
```
