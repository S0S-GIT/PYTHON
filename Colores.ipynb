```python
import matplotlib.pyplot as plt
import numpy as np
imagen = np.linspace(0, 1, 1000)
imagen = np.tile(imagen, (300, 1))
plt.imshow(imagen, cmap="copper_r")
plt.axis("off")
plt.show()
```

```python
imagen = np.zeros((300,300))
imagen[100:200,100:200] = 1
plt.imshow(imagen, cmap="copper_r")
plt.axis("off")
plt.show()
```
