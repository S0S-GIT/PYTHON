```python
import cv2
image = cv2.imread("/content/B.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
text = "HOLA"
position = (50,50)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
cv2.putText(image, text, position, cv2.FONT_HERSHEY_COMPLEX, 1, (0, 0, 255), 2, cv2.LINE_AA)
plt.imshow(image)
plt.axis("off")
plt.show()
```
