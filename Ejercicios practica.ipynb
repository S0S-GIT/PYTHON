ACT.1 Escribe un programa que genere 5 numero aleatorios entre el 1 y 10

Por cada numero:
- Si es mayor o igual que 5, suma un punto
- Si es menor que 5, no suma puntos

Al terminar muestre la cantidad de puntos obtenidos
