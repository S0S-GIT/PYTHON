```python
def ver(num,sec):
  if num<sec:
    return"El número secreto es mayor"
  elif num>sec:
    return"El número secreto es menor"
  else:
    return"¡Correcto!"
sec=7
print("Adivina el numero")
print("Tienes 3 intentos")
for intento in range(1,4):
  num = int(input("Intento" + str(intento) + ": "))
  res = ver(num, sec)
  print(res)
  if num == sec:
    break
```
