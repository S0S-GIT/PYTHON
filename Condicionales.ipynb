_Ejercicio: Elabora un script el cual tenga_
1. Una función llamada com_f(num1,num2) y reciba parametros de entrada.
2. Debe imprimir la relación entre los dos numeros:Menor que, Mayor que o igual que.
3. La función no retorna ningun valor.
Resultado esperado: El número *_X_* es menor que *_Y_*

```python
def con_f(num1, num2):
  if num1 < num2:
    print(f"El número {num1} es menor que {num2}")
  elif num1>num2:
    print(f"El número {num2} es menor que {num1}")
  else:
    print(f"El número {num1} es igual {num2}")
```

```python
num1 =int(input("Ingresa un numero entero: "))
num2 =int(input("Ingresa el segundo numero entero: "))
con_f(num1,num2)
```
